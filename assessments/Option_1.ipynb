## Abstract

This study examines how introducing different spaceship patterns in different starting rows disrupts the Gosper Glider Gun (GGG) in Conway's Game of Life. Three spaceships types - Lightweight(LWSS), Middleweight (MWSS), and Heavyweight(HWSS) - were placed at five starting row positions (2, 4, 6, 8, and 20) within a Python-simulated 2D universe and observed over 210 rounds. Results revealed that collision outcomes depend heavily on both spaceship type and starting position. Spaceships placed far from the GGG(row 20) typically annihilated a single glider before the system recovered, while those placed closer triggered destructive chain reactions that dismantled the GGG entirely. Also, heavier spaceships generally caused greater and more prolonged disruption. Most close-range collisions had not stabilised by round 210, ultimately resolving into simple still-life patterns such as blocks, beehives, and blinkers after extended simulation.


## Introduction
  
<!-- Introduction about Conway's Game of Life -->
Conway's Game of Life is a mathematical simulation created by mathematician John Korton Conway in 1970. It is a type of cellular automaton where cells on a two-dimensional grid change between alive and dead states according to a simple set of rules. Although the rules are basic, the simulation can generate complex and unpredictable patterns, including stable structures and moving formations [1][2].

The Game of Life is widely studied in computer science and mathematics because it demonstrates how complex behavior can emerge from simple logical instructions. It is also used as a learning tool from programming, algorithms, and problem-solving due to its interactive and visual nature[3].

<!-- Gosper Glider Gun and what that does and what it represents -->
The Gosper Glider Gun(GGG) is one of the most famous patterns in Conway's Game of Life. Created by Bill Gosper in 1970, it was the first pattern discovered that could continuously produce moving patterns called gliders [4].  The GGG consist of two Queen Bee Shuttles and two eaters. The Queen Bee Shuttles is an oscillator shuttle with a period of 30, meaning it returns to its original state after 30 generations. It works by sending sparks back and forth between stabilizing structures, creating a continuous rhythmic motion. In contrast, a block is a stable pattern that does not change over time and is designed to absorb moving objects such as gliders without being destroyed. Blocks are useful because they remove unwanted activity and help control signals in larger Life constructions. Together, these patterns demonstrate how the Game of Life can support organized behavior, with the Queen Bee Shuttle generating activity and the Eater safely stopping or controlling it [5]. A glider is a small arrangement of cells that moves diagonally across the grid while changing shape over time. The GGG repeatedly creates these gliders automatically, allowing the pattern to continue indefinitely [6].

The GGG demonstrates self-sustaining behavior, infinite growth, and information transmission. The emitted gliders can travel indefinitely across the board and interact with other patterns. These interactions can create logic gates and computational structures. Researchers later used glider streams to build AND gates, OR gates, NOT gates, memory systems, and ultimately systems equivalent to a universal computer. This is one reason the Game of Life is considered Turing complete---meaning it can theoretically compute anything a normal computer can [4].

The GGG mattered historically because it disproved an early conjecture by Conway that all finite patterns must eventually stabilize or die out. Gosper's discovery proved that a finite pattern could generate unlimited growth forever. Its importance in computer science helped establish cellular automata as computationally universal, in artifical life showed how lifelike behavior can emerge algorithmically, in complexity theory became a classic example if emergent systems, and in mathematics demonstrated unexpected complexity from simple systems [3] [4] [7].

## Code for Conway's Game of Life

```python
#### -------- INSERT CODE HERE ----------

import numpy as np #import the numerical Python library, numpy. Changing the referenced library to 'np' is solely for convenience
import matplotlib.pyplot as plt #as per the above, much easier to write over and over again
from IPython.display import clear_output
import time

UNIVERSE_HEIGHT = 50
UNIVERSE_WIDTH = 100

# Implement the Gosper Glider Gun
GosperGliderGun = np.zeros((9,36), dtype=int)

GosperGliderGun[4:6, 0:2] = 1
GosperGliderGun[2:4, 34:36] = 1
GosperGliderGun[2:5, 20:22] = 1

rows = [2, 2, 3, 4, 5, 6, 7, 8, 8, 5, 3, 7, 4, 5, 6, 5, 1, 5, 0, 1, 5, 6]
cols = [12, 13, 11, 10, 10, 10, 11, 12, 13, 14, 15, 15, 16, 16, 16, 17, 22, 22, 24, 24, 24, 24]
GosperGliderGun[rows, cols] = 1

# Implement the Lightweight Spaceship (LWSS)
LWSS = np.zeros((6,8),dtype=int)
rows = [1, 1, 2, 3, 4, 4, 4, 4, 3]
cols = [2, 5, 1, 1, 1, 2, 3, 4, 5]
LWSS[rows, cols] = 1

# Implement the Middleweight Spaceship (MWSS)
MWSS = np.zeros((6,8),dtype=int)
rows = [3, 4, 5, 2, 5, 5, 1, 5, 5, 5, 2, 4]
cols = [1, 1, 1, 2, 2, 3, 4, 4, 5, 6, 7, 7]
MWSS[rows, cols] = 1

# Implement the Heavyweight Spaceship (HWSS)
HWSS = np.zeros((6,8),dtype=int)
rows = [2, 3, 4, 1, 4, 4, 4, 4, 4, 3]
cols = [1, 1, 1, 2, 2, 3, 4, 5, 6, 7]
HWSS[rows, cols] = 1

spaceShips = {"LWSS": LWSS, "MWSS": MWSS, "HWSS": HWSS}

def search_each_cell(total_rows,total_cols,space):
    new_space = np.zeros((total_rows,total_cols),dtype=int)
    # 1) First, we need to start moving through each cell of our 'space'.
    #    To do this, we will use two nested 'for' loops
    for row in range(total_rows):
        for col in range(total_cols):
            # So 'row' and 'col' define our current cell index.
            # We now need to search a neighbourhood defined as 1 cell distance around this position
            # We thus need two more nested for loops. When searching this neighbouring space, we want
            # to count the number of 1's. Thus, we also need a variable that we can increment by 1
            # everytime we find a value of 1. Let us call this integer variable count

            count = 0
            for row2 in range(row-1,row+2): # See here that we can define a start and end to our 'range'
                for col2 in range(col-1,col+2):
                     # We need to check if our new position, defined by [row2,col2] is off the board
                    if (row2<0) or (row2>=total_rows) or (col<0) or (col2>=total_cols):
                        # Do nothing
                        pass
                    elif row2 == row and col2 == col:
                        # Do nothing, it's the cell we already have!
                        pass
                    # If we are not off the board or in the same cell as our starting point...
                    # We can check if this new space has a value of 1. If it does, lets count it
                    else:
                        if space[row2,col2]>0:
                            count=count+1
            # Here you need to introduce conditional statements that act on the value of 'count'
            # Read through the narrative provided above and remember to obey the spacing rules
            if space[row,col] > 0:
                if count < 2:
                    new_space[row,col] = 0
                elif 2<=count<=3:
                    new_space[row,col] = 1
                elif count > 3:
                    new_space[row,col] = 0
            elif space[row,col] == 0:
                if count == 3:
                    new_space[row,col] = 1

    return new_space

# Menu system to choose between the two options of Simulation.
print("Choose one of the options below:")
print("1 - Show outputs of all the simulations for various spaceships after every 30 rounds.")
print("2 - Show a single simulation")
choice = input("Option:")
if choice == "1":
    # If you want to print out what the Universe simulation will look like for various spaceships
    # starting at various rows and showing what each simulation looks like after every 30 rounds.
    # use this code below instead:

    ROUND_INTERVAL = 30

    # Go through all the spaceships one by one
    for key in spaceShips:
        # Go through all the different starting rows one by one
        for startingRow in (2, 4, 6, 8, 20):
            # Set up the Universe
            Universe = np.zeros((UNIVERSE_HEIGHT, UNIVERSE_WIDTH), dtype=int)
            # Place the GosperGliderGun and Spaceship into the Universe
            Universe[0:9, 0:36] = GosperGliderGun
            Universe[startingRow:startingRow+6, 80:88] = spaceShips[key]

            # Creating a 2x4 groups of subplots to plot Simulation after 30, 60, 90, 120, 150, 180 and 210 rounds.
            fig, axes = plt.subplots(2, 4, figsize=(16, 6))
            fig.suptitle(f"Spaceship Used: {key}")
            for index, ax in enumerate(axes.flatten()):
                ax.imshow(Universe, cmap='binary') # The cmap, or colour map, gives us a black and white board.
                ax.set_title(f"Starting Row: {startingRow}\nRound: {index * ROUND_INTERVAL}", fontsize=10)
                for _ in range(ROUND_INTERVAL):
                    Universe = search_each_cell(UNIVERSE_HEIGHT, UNIVERSE_WIDTH, Universe)
            plt.show()
        # Keep a gap between each Simulation using the print() function below
        print("\n" * 3)
elif choice == "2":
    # If you want to animate for a single scenario, use this code below instead:

    FRAME_RATE = 30          # This determines the no. of rounds that will happen every second. Higher FRAME_RATE means faster simulation
    ROUNDS = 360             # The total number of rounds the simulation will take place. We will look at 30, 60, 90, 120, 150, 180, and 210.
    STARTING_ROW = 20        # Change the values between 2 and 20 to see the collision between GosperGliderGun, its gliders and spaceships
    SPACESHIP = "HWSS"       # Change between LWSS, MWSS and HWSS

    # Set up the Universe
    Universe = np.zeros((UNIVERSE_HEIGHT, UNIVERSE_WIDTH), dtype=int)

    # Place the GosperGliderGun and LWSS into the Universe
    Universe[0:9, 0:36] = GosperGliderGun
    if SPACESHIP == "LWSS":
        Universe[STARTING_ROW:STARTING_ROW+6, 80:88] = LWSS
    elif SPACESHIP == "MWSS":
        Universe[STARTING_ROW:STARTING_ROW+6, 80:88] = MWSS
    elif SPACESHIP == "HWSS":
        Universe[STARTING_ROW:STARTING_ROW+6, 80:88] = HWSS

    for x in range(ROUNDS):
        clear_output(wait=True)
        Universe_new = search_each_cell(UNIVERSE_HEIGHT, UNIVERSE_WIDTH, Universe)
        Universe = Universe_new
        plt.imshow(Universe_new, cmap='binary') #The cmap, or colour map, gives us a black and white board.
        plt.title('Universe simulation')
        plt.show()
        time.sleep(1 / FRAME_RATE)
else:
    print("Wrong choice! Please choose one of the options below: " )

#### ------------------------------------
```
